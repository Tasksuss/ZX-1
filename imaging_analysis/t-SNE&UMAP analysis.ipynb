```python
# ============================================================
# Block 0: Load saved projection results from analysis folder
# Assumes the previous notebook already saved:
#   - framewise_mode_coefficients.csv.gz
#   - optionally other summary files
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------------- USER CONFIG ----------------
RESULTS_ROOT = "/content/drive/My Drive/Centre_align_cropped_output"

# Saved files from the previous notebook
PROJ_FILE = os.path.join(RESULTS_ROOT, "framewise_mode_coefficients.csv.gz")
# optional extras if you want them later:
CLIP_FEATURE_FILE = os.path.join(RESULTS_ROOT, "clip_level_features.csv")
CORR_FILE = os.path.join(RESULTS_ROOT, "correlation_ready_clip_features.csv")

# ---------------- CHECK PATHS ----------------
if not os.path.exists(RESULTS_ROOT):
    raise FileNotFoundError(f"RESULTS_ROOT not found:\n{RESULTS_ROOT}")

if not os.path.exists(PROJ_FILE):
    raise FileNotFoundError(
        f"Projection file not found:\n{PROJ_FILE}\n"
        "Check the output folder and filename."
    )

print("Loading projection table from:")
print(PROJ_FILE)

# ---------------- LOAD MAIN FRAMEWISE COEFFICIENT TABLE ----------------
proj_df = pd.read_csv(PROJ_FILE, compression="gzip")

print("\nLoaded proj_df successfully.")
print("Shape:", proj_df.shape)

print("\nColumns:")
print(proj_df.columns.tolist()[:30], "..." if len(proj_df.columns) > 30 else "")

# Basic sanity checks
required_cols = ["rel_path", "subject", "word", "frame_idx"]
missing = [c for c in required_cols if c not in proj_df.columns]
if len(missing) > 0:
    raise ValueError(f"proj_df is missing required columns: {missing}")

mode_cols = [c for c in proj_df.columns if c.startswith("mode_")]
print(f"\nNumber of mode columns found: {len(mode_cols)}")
print("First few mode columns:", mode_cols[:10])

if len(mode_cols) == 0:
    raise ValueError("No mode_* columns found in proj_df.")

# infer n_keep from saved table
n_keep = len(mode_cols)
print("Inferred n_keep =", n_keep)

print("\nUnique subjects:")
print(sorted(proj_df["subject"].dropna().astype(str).unique()))

print("\nUnique words:")
print(sorted(proj_df["word"].dropna().astype(str).unique()))

print("\nPreview:")
display(proj_df.head())
```
